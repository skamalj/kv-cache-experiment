# ModernBERT-JEPA decider: LoRA encoder and a transformer predictor

Run 11 (`05_jepa_point.ipynb`) trained all of ModernBERT-base with the **JEPA pointer** and lifted JevBench's public items from 0.394 (run 8 B) to 0.493. Its predictor was a small MLP on the `[MASK]` vector. This run changes two things:

- **LoRA encoder.** ModernBERT's weights are frozen; rank-`LORA_RANK` adapters on its attention and MLP projections train (with the MLM transform, the label head, the predictor and τ). The EMA target is the same frozen ModernBERT with an **EMA copy of the adapters**.
- **A transformer predictor (I-JEPA style).** A narrow transformer (`PRED_LAYERS` layers, width `PRED_WIDTH`) reads the encoder's **whole output** (state, question, every option's label token), and is read at `[MASK]`: a second, question-focused read before ẑ lands. Its output layer starts at zero, so training starts from run 11's ẑ = `[MASK]` vector.

Unchanged from run 11: the input `[CLS] question [SEP] state [SEP] A: option … Answer: [MASK]`; the targets (the EMA encoder's vectors at the option label tokens); the JEPA loss (ẑ onto the gold option's label vector); the pointer (softmax of cos(ẑ, v_k) / τ over the question's own labels, every other label 0); the 38-way head; C2's decision averaging head and pointer.

**Arms** (tasksource and our data mixed, `ARM_EPOCHS` passes, 1 by default):
- **L3:** LoRA + transformer predictor (first).
- **L2:** LoRA + MLP predictor (run 11's predictor), so the predictor's effect is measured with the encoder held the same.
- **L0 (control, chosen with `ARMS`):** the same LoRA encoder and head with **no JEPA** (no predictor loss, no target, no pointer), so the effect of JEPA itself is measured on the same hardware, data and schedule.

`SEED` changes training only (initialisation, batch order); `DATA_SEED` fixes the tasksource sample, so seed repeats train on the same data.

Each arm is scored at each quarter (1,000 of its training questions, our dev, JevBench), at the end on Typed Decisions' train split, and then with **JevBench's own CLI** on the 231 public items, served from this kernel (as in `06_jevbench_cli.ipynb`): the CLI's numbers, and our Intelligence estimate from the board's published rules, for the averaged decision and for the head alone.

**Checkpoints** every `CKPT_EVERY` steps and after each quarter, to `CKPT_DIR`; after a crash, restart the kernel and run all cells. A finished arm's results (including its CLI scores) are read back; its final weights (adapters, EMA adapters, heads, predictor) are saved small, since the frozen ModernBERT comes from the hub.

ModernBERT-large: set `ENCODER` to `answerdotai/ModernBERT-large`; gradient checkpointing then turns on.

## Device and log

One or more GPUs: with two, every batch is split across both. Mixed precision is bf16 where the GPU has it and fp16 with loss scaling otherwise (a T4).

```python
DEVICE = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
N_GPU = torch.cuda.device_count()
AMP = DEVICE.type == "cuda"
# bf16 only with bf16 hardware (compute capability 8.0+: A100, G4); a T4 reports bf16 as
# supported but only emulates it, very slowly, so it uses fp16 with loss scaling
BF16_HW = AMP and torch.cuda.get_device_capability(0)[0] >= 8
AMP_DTYPE = torch.bfloat16 if BF16_HW else torch.float16


def amp():
    return torch.autocast("cuda", dtype=AMP_DTYPE, enabled=AMP)


def new_scaler():
    """Loss scaling, needed for fp16 only (bf16 has fp32's range)."""
    return torch.amp.GradScaler("cuda", enabled=AMP and AMP_DTYPE == torch.float16)


def spread(module):
    """With several GPUs, each batch is split across them (data parallel)."""
    return nn.DataParallel(module) if N_GPU > 1 else module


def log(message):
    print(f"[{time.strftime('%H:%M:%S')}] {message}", flush=True)


gpu = f"{N_GPU} x {torch.cuda.get_device_name()}" if AMP else "CPU"
log(f"torch {torch.__version__} | transformers {transformers.__version__} | {gpu} | "
    f"{str(AMP_DTYPE).split('.')[-1] if AMP else 'fp32'} | {ENCODER}")
```

## One item per question

Every question becomes: question text, state text, option texts and a gold distribution over the options. `noul` options are the criteria's `true` / `false` descriptions (or nothing), shown as `yes` / `no`; `choice` options are `name - description`, shown as `A`, `B`, …; `score` levels are shown as `0`, `1`, …

```python
def state_text(state):
    return state if isinstance(state, str) else json.dumps(state, ensure_ascii=False)


def option_texts(q):
    """-> (option names, option texts) for one System One question."""
    kind, crit = q["type"], q.get("criteria")
    if kind == "noul":
        crit = crit or {}
        return ["true", "false"], [crit.get("true") or "", crit.get("false") or ""]
    if kind == "choice":
        names = list(crit)
        return names, [f"{n} - {d}" if d else n for n, d in crit.items()]
    return [str(i) for i in range(len(crit))], [str(level) for level in crit]


def make_item(kind, question, state, q, gold, source, family=""):
    names, texts = option_texts(q)
    total = sum(gold)
    return {"type": kind, "question": question or "", "state": state_text(state), "names": names,
            "texts": texts, "gold": [g / total for g in gold], "source": source, "family": family}
```

## Our training and dev data

```python
def load_split(split):
    path = hf_hub_download(DATASET, f"data/{split}.jsonl", repo_type="dataset", token=HF_TOKEN)
    items = []
    for line in open(path, encoding="utf-8"):
        rec = json.loads(line)
        for qid, q in rec["questions"].items():
            probs = rec["gold"][qid]["probabilities"]
            names, _ = option_texts(q)
            if len(names) > MAX_OPTIONS:
                continue
            gold = [float(probs[n]) for n in names]
            items.append(make_item(q["type"], q.get("instructions"), rec["state"], q, gold, split))
    return items


train_all, dev_items = load_split("train"), load_split("dev")
train_items = (random.Random(SEED).sample(train_all, min(TRAIN_QUESTIONS, len(train_all)))
               if TRAIN_QUESTIONS else train_all)
print(f"train {len(train_items)} of {len(train_all)} questions "
      f"{dict(Counter(it['type'] for it in train_items))} | dev {len(dev_items)}")
```

## JevBench public items (test only, never trained on)

One question per item and one expected label, so the gold is one-hot.

```python
def load_jevbench(split):
    url = (f"https://raw.githubusercontent.com/fstandhartinger/jevbench/{JEVBENCH_COMMIT}"
           f"/datasets/public/{split}.jsonl")
    items = []
    for line in urllib.request.urlopen(url).read().decode("utf-8").splitlines():
        rec = json.loads(line)
        q = rec["question"]
        names, _ = option_texts(q)
        if q["type"] == "noul":
            target = "true" if rec["expected"] == "yes" else "false"
        else:
            target = str(rec["expected"])
        gold = [1.0 if n == target else 0.0 for n in names]
        assert sum(gold) == 1, rec["id"]
        items.append(make_item(q["type"], q["instructions"], rec["state"], q, gold,
                               split, rec["family"]))
    return items


jev_items = [it for split in ("original", "easy", "hard") for it in load_jevbench(split)]
print(f"JevBench public: {len(jev_items)} items {dict(Counter(it['source'] for it in jev_items))} "
      f"{dict(Counter(it['type'] for it in jev_items))}")
```

## Stage 1 data: public typed decisions, filtered

A guard set of word 8-grams from every test state we care about (JevBench's public items, Typed Decisions' train and test splits). Then the tasksource stream, shuffled: rows whose `license_use` is commercial, with 2–26 options (2–10 levels for `score`), at most `TS_PER_SOURCE` per source, and no 8-gram shared with the guard set.

```python
WORD = re.compile(r"\w+")


def ngrams(text, n=OVERLAP_N):
    words = WORD.findall(text.lower())
    return {" ".join(words[i:i + n]) for i in range(len(words) - n + 1)}


guard = set()
for it in jev_items:
    guard |= ngrams(it["state"])
for split in ("train", "test"):
    path = hf_hub_download("LocalLLaMA/typed-decisions", f"all/{split}-00000-of-00001.parquet",
                           repo_type="dataset")
    for state in pd.read_parquet(path)["state"]:
        guard |= ngrams(state_text(json.loads(state)))
print(f"guard set: {len(guard):,} 8-grams")
```

## Tasksource sample (cached)

The tasksource stream, shuffled: rows whose `license_use` is commercial, with 2–26 options (2–10 levels for `score`), at most `TS_PER_SOURCE` per source, and no 8-gram shared with the guard set. The sample is saved in `CKPT_DIR`; a rerun reads it from there (or from run 11's or run 12's folder, the same sample).

```python
def ts_item(row):
    """A tasksource row -> one item, or None if it does not fit."""
    kind, opts, target = row["kind"], row["options"], row["target"]
    if kind == "noul":
        q, gold = {"type": "noul", "criteria": None}, [target[0], 1 - target[0]]
    elif kind == "choice" and 2 <= len(opts) <= MAX_OPTIONS and len(set(opts)) == len(opts):
        q, gold = {"type": "choice", "criteria": {o: None for o in opts}}, list(target)
    elif kind == "score" and 2 <= len(opts) <= 10:
        q, gold = {"type": "score", "criteria": list(opts)}, list(target)
    else:
        return None
    if sum(gold) <= 0 or len(gold) != len(option_texts(q)[0]):
        return None
    return make_item(kind, row["question"], row["state"], q, gold, "tasksource", row["source"])


def sample_tasksource():
    items, per_source, dropped = [], Counter(), Counter()
    stream = load_dataset(TASKSOURCE, "default", split="train", streaming=True, token=HF_TOKEN)
    for row in stream.shuffle(seed=DATA_SEED, buffer_size=TS_BUFFER):
        if row["license_use"] != "commercial":
            dropped["licence"] += 1
            continue
        if per_source[row["source"]] >= TS_PER_SOURCE:
            continue
        item = ts_item(row)
        if item is None:
            dropped["shape"] += 1
            continue
        if ngrams(item["state"]) & guard:
            dropped["overlap"] += 1
            continue
        items.append(item)
        per_source[row["source"]] += 1
        if len(items) >= TS_QUESTIONS:
            break
    print(f"sampled from {len(per_source)} sources | dropped {dict(dropped)}")
    return items


ts_name = f"tasksource_{TS_QUESTIONS}_{TS_PER_SOURCE}_{DATA_SEED}.json"
ts_cache = next((os.path.join(d, ts_name) for d in TS_CACHE_DIRS
                 if os.path.exists(os.path.join(d, ts_name))), os.path.join(CKPT_DIR, ts_name))
if os.path.exists(ts_cache):
    ts_items = json.load(open(ts_cache, encoding="utf-8"))
    print(f"read from {ts_cache}")
else:
    ts_items = sample_tasksource()
    json.dump(ts_items, open(ts_cache, "w", encoding="utf-8"))
print(f"tasksource: {len(ts_items)} questions {dict(Counter(it['type'] for it in ts_items))}")
```

## Label tokens

The 38 labels are A–Z, yes / no and 0–9. With the leading space they have after `Answer:`, each must be one token of ModernBERT's vocabulary: the head starts from those tokens' pretrained output rows. At the start of an option line (`A: …`) a label must keep its own tokens, so its position can be found; both are checked here.

```python
tok = AutoTokenizer.from_pretrained(ENCODER)
CLS, SEP, PAD, MASK = tok.cls_token_id, tok.sep_token_id, tok.pad_token_id, tok.mask_token_id
LABELS = {"choice": [chr(65 + i) for i in range(MAX_OPTIONS)],
          "noul": ["yes", "no"],
          "score": [str(i) for i in range(10)]}


def ids(text):
    return tok(text, add_special_tokens=False)["input_ids"]


def one_token(label):
    found = ids(" " + label)
    assert len(found) == 1, f"label {label!r} is {len(found)} tokens"
    return found[0]


LABEL_IDS = {kind: [one_token(lab) for lab in labs] for kind, labs in LABELS.items()}
HEAD_TOKENS = LABEL_IDS["choice"] + LABEL_IDS["noul"] + LABEL_IDS["score"]   # the 38 head outputs
HEAD_SLOTS = {"choice": list(range(26)), "noul": [26, 27], "score": list(range(28, 38))}
for labs in LABELS.values():
    for lab in labs:
        assert ids(f"{lab}: x")[:len(ids(lab))] == ids(lab), f"label {lab!r} merges with ':'"
print({k: v[:4] for k, v in LABEL_IDS.items()}, "| head outputs", len(HEAD_TOKENS))
```

## One sequence per question

`[CLS] question [SEP] state [SEP] A: option … [SEP] Answer: [MASK] [SEP]`. The question and the options are kept whole (each option cut at `OPTION_MAX_TOKENS`); if the sequence is too long, the state's end is cut, and the cut is counted. Also returned: the position of each option's label token (the pointer's targets). In training, choice options are shuffled, so a letter never carries a fixed answer.

```python
def shuffled(item, rng):
    """A copy of a choice item with its options in a random order (gold moved with them)."""
    if item["type"] != "choice" or rng is None:
        return item
    order = list(range(len(item["names"])))
    rng.shuffle(order)
    return {**item, "names": [item["names"][i] for i in order],
            "texts": [item["texts"][i] for i in order], "gold": [item["gold"][i] for i in order]}


def sequence(item, max_len):
    """-> (token ids with [MASK], mask position, label positions, state was cut)."""
    labels = LABELS[item["type"]]
    tail, at = [], []
    for label, text in zip(labels, item["texts"]):
        line = f"{label}: {text}".strip() if text else label
        at.append(len(tail) + len(ids(label)) - 1)        # the label's (last) token
        tail += ids(line)[:OPTION_MAX_TOKENS] + [SEP]
    tail += ids("Answer:")
    q = ids(item["question"])[:256]
    s = ids(item["state"])
    room = max_len - len(q) - len(tail) - 5
    cut = len(s) > room
    s = s[:max(room, 0)]
    head = [CLS] + q + [SEP] + s + [SEP]
    seq = head + tail
    pos = len(seq)
    seq += [MASK, SEP]
    return seq, pos, [len(head) + a for a in at], cut


def batch_tensors(items, max_len):
    rows = [sequence(it, max_len) for it in items]
    width = max(len(r[0]) for r in rows)
    k = max(len(r[2]) for r in rows)
    input_ids, att, pos, label_pos = [], [], [], []
    for seq, p, at, _ in rows:
        input_ids.append(seq + [PAD] * (width - len(seq)))
        att.append([1] * len(seq) + [0] * (width - len(seq)))
        pos.append(p)
        label_pos.append(at + [0] * (k - len(at)))        # padding points at [CLS]; masked out
    as_t = lambda x: torch.tensor(x, device=DEVICE)
    return as_t(input_ids), as_t(att), as_t(pos), as_t(label_pos), sum(r[3] for r in rows)
```

## The model

- **Encoder:** ModernBERT, frozen, with LoRA adapters (trained) on `Wqkv`, `Wo` and `Wi` of every layer.
- **MLM transform** (trained) and **label head** (trained): 38 outputs (A–Z, yes / no, 0–9), started from the pretrained output rows of those tokens; each question keeps only its own labels.
- **Predictor** (trained), ẑ = `[MASK]` vector + predictor output:
  - `transformer`: the encoder's whole output, projected to `PRED_WIDTH`, through `PRED_LAYERS` pre-norm transformer layers (padding masked), read at `[MASK]`, projected back; the last projection starts at zero.
  - `mlp`: run 11's MLP on the `[MASK]` vector.
- **Target encoder:** the same frozen ModernBERT with an EMA copy of the adapters (no gradients, always in eval mode); its vectors at the option label tokens are the targets and the pointer's options.
- **Temperature τ** (trained) for the pointer.

```python
TYPES = ["noul", "choice", "score"]


class EncoderOut(nn.Module):
    """An encoder's output at every position ([rows, length, hidden]), or at given positions
    ([rows] or [rows, k]); split across GPUs by spread()."""

    def __init__(self, encoder):
        super().__init__()
        self.enc = encoder

    def forward(self, input_ids, att, pos=None):
        with amp():                                   # each GPU's copy runs in mixed precision
            h = self.enc(input_ids=input_ids, attention_mask=att).last_hidden_state
        if pos is None:
            return h
        rows = torch.arange(len(pos), device=h.device)
        return h[rows, pos] if pos.dim() == 1 else h[rows[:, None], pos]


class MLPPredictor(nn.Module):
    """Run 11's predictor: an MLP on the [MASK] vector."""

    def __init__(self, hidden):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(hidden, 2 * hidden), nn.GELU(),
                                 nn.LayerNorm(2 * hidden), nn.Linear(2 * hidden, hidden))

    def forward(self, h, att, pos):
        return self.net(h[torch.arange(len(pos), device=h.device), pos])


class TransformerPredictor(nn.Module):
    """I-JEPA style: a narrow transformer over the encoder's whole output, read at [MASK]."""

    def __init__(self, hidden):
        super().__init__()
        self.inp = nn.Linear(hidden, PRED_WIDTH)
        layer = nn.TransformerEncoderLayer(PRED_WIDTH, PRED_HEADS, 4 * PRED_WIDTH, dropout=0.0,
                                           batch_first=True, norm_first=True)
        self.layers = nn.TransformerEncoder(layer, PRED_LAYERS, enable_nested_tensor=False)
        self.norm = nn.LayerNorm(PRED_WIDTH)
        self.out = nn.Linear(PRED_WIDTH, hidden)
        nn.init.zeros_(self.out.weight)               # starts as run 11's z = [MASK] vector
        nn.init.zeros_(self.out.bias)

    def forward(self, h, att, pos):
        x = self.layers(self.inp(h), src_key_padding_mask=att == 0)
        return self.out(self.norm(x[torch.arange(len(pos), device=h.device), pos]))


class MaskedDecider(nn.Module):
    def __init__(self, predictor="transformer"):
        super().__init__()
        mlm = AutoModelForMaskedLM.from_pretrained(ENCODER, attn_implementation="sdpa")
        hidden = mlm.config.hidden_size
        self.head = mlm.head                          # the MLM transform
        self.label_head = nn.Linear(hidden, len(HEAD_TOKENS))
        with torch.no_grad():                         # start from the pretrained output rows
            self.label_head.weight.copy_(mlm.decoder.weight[HEAD_TOKENS])
            if mlm.decoder.bias is not None:
                self.label_head.bias.copy_(mlm.decoder.bias[HEAD_TOKENS])
        base = mlm.model
        del mlm
        if GRAD_CHECKPOINTING:                        # only active in training mode
            base.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
        config = LoraConfig(r=LORA_RANK, lora_alpha=LORA_RANK, lora_dropout=0.0,
                            target_modules=["Wqkv", "Wo", "Wi"])
        self.encoder = get_peft_model(base, config)   # ModernBERT frozen; the adapters train
        self.predictor = (TransformerPredictor if predictor == "transformer"
                          else MLPPredictor)(hidden)
        self.log_tau = nn.Parameter(torch.tensor(math.log(TAU_START)))
        self.target = copy.deepcopy(self.encoder).requires_grad_(False)
        self.online = spread(EncoderOut(self.encoder))
        self.target_at = spread(EncoderOut(self.target))

    def label_logits(self, vectors, items):
        """The 38-way head at [MASK]; each question keeps its own labels -> [items, options]."""
        all38 = self.label_head(self.head(vectors)).float()
        width = max(len(it["names"]) for it in items)
        out = torch.full((len(items), width), -1e4, device=vectors.device)
        for i, it in enumerate(items):
            k = len(it["names"])
            out[i, :k] = all38[i, HEAD_SLOTS[it["type"]][:k]]
        return out

    def tau(self):
        return self.log_tau.exp().clamp(min=0.01)

    def train(self, mode=True):
        """Train the online parts; the EMA target always stays in eval mode."""
        super().train(mode)
        self.target.eval()
        return self

    @torch.no_grad()
    def target_vectors(self, input_ids, att, label_pos):
        return self.target_at(input_ids, att, label_pos)

    def final_state(self):
        """What a trained arm needs beyond the hub's ModernBERT: adapters, EMA adapters, heads,
        predictor, temperature."""
        def wanted(key):
            top = key.split(".")[0]
            if top in ("online", "target_at"):        # the same tensors under the GPU wrappers
                return False
            return "lora_" in key or top not in ("encoder", "target")

        return {k: v.detach().cpu() for k, v in self.state_dict().items() if wanted(k)}
```

## Scores and losses for a batch

Scores per readout: the head's, and with the pointer on, cos(ẑ, v_k) / τ over the question's own labels. The predictor reads the encoder's whole output (`transformer`) or the `[MASK]` vector (`mlp`). Losses: KL to the gold for the head and for the pointer; the ordinal error for `score` (head); JEPA, 1 − cos(ẑ, v_k) weighted by the gold distribution. The probabilities: the head's, the pointer's, and both averaged (C2's decision).

```python
def forward(model, items, max_len, pointer):
    input_ids, att, pos, label_pos, cut = batch_tensors(items, max_len)
    h = model.online(input_ids, att)                                       # [items, length, hidden]
    v = h[torch.arange(len(items), device=h.device), pos]
    scores, extra = {"head": model.label_logits(v, items)}, {}
    if pointer:
        targets = model.target_vectors(input_ids, att, label_pos).float()   # [items, k, hidden]
        z = (v + model.predictor(h, att, pos)).float()
        cos = F.cosine_similarity(z.unsqueeze(1), targets, dim=-1)
        k = torch.tensor([len(it["names"]) for it in items], device=cos.device)
        valid = torch.arange(cos.shape[1], device=cos.device)[None] < k[:, None]
        scores["point"] = (cos / model.tau()).masked_fill(~valid, -1e4)
        extra = {"z": z, "targets": targets, "cos": cos, "valid": valid}
    return scores, extra, cut


def gold_matrix(items, width):
    g = torch.zeros(len(items), width, device=DEVICE)
    for i, it in enumerate(items):
        g[i, :len(it["gold"])] = torch.tensor(it["gold"], device=DEVICE)
    return g


def kl(g, scores):
    logp = F.log_softmax(scores, dim=1)
    return (g * (g.clamp(min=1e-12).log() - logp)).sum(1).mean(), logp


def losses(scores, extra, items):
    g = gold_matrix(items, scores["head"].shape[1])
    dec, logp = kl(g, scores["head"])
    parts = {"dec": dec}
    is_score = torch.tensor([it["type"] == "score" for it in items], device=DEVICE)
    if is_score.any():
        levels = torch.arange(g.shape[1], device=DEVICE, dtype=torch.float32)
        n = torch.tensor([len(it["gold"]) - 1 for it in items], device=DEVICE).clamp(min=1)
        gap = ((logp.exp() - g) * levels).sum(1).abs() / n
        parts["ord"] = gap[is_score].mean()
    if "point" in scores:
        parts["point"] = kl(g, scores["point"])[0]
        parts["jepa"] = (g * (1 - extra["cos"])).sum(1).mean()
    weights = {"dec": 1.0, "ord": LAMBDA_ORD, "point": LAMBDA_POINT, "jepa": LAMBDA_JEPA}
    total = sum(weights[k] * v for k, v in parts.items())
    return total, {k: v.item() for k, v in parts.items()}


def readouts(scores):
    """Probabilities per readout: the head, the pointer, and both averaged."""
    probs = {name: F.softmax(s, dim=1) for name, s in scores.items()}
    if "point" in probs:
        probs["both"] = (probs["head"] + probs["point"]) / 2
    return probs
```

## The measures

JevBench's: accuracy (argmax), multi-class Brier, top-label ECE over 10 equal-width bins (empty bins left out), ordinal MAE for `score`; and KL for our soft dev gold.

```python
def measure(rows):
    n = len(rows)
    bins = defaultdict(list)
    for r in rows:
        bins[min(int(r["conf"] * 10), 9)].append(r)
    ece = 0.0
    for b in bins.values():
        right = sum(x["ok"] for x in b) / len(b)
        sure = sum(x["conf"] for x in b) / len(b)
        ece += len(b) / n * abs(right - sure)
    out = {"n": n, "acc": round(sum(r["ok"] for r in rows) / n, 4),
           "brier": round(statistics.mean(r["brier"] for r in rows), 4),
           "ece": round(ece, 4), "kl": round(statistics.mean(r["kl"] for r in rows), 4)}
    mae = [r["mae"] for r in rows if r["mae"] is not None]
    if mae:
        out["ordinal_mae"] = round(statistics.mean(mae), 3)
    return out


def row_of(it, p):
    g = it["gold"]
    top, want = max(range(len(p)), key=p.__getitem__), max(range(len(g)), key=g.__getitem__)
    mae = abs(sum(i * x for i, x in enumerate(p)) - want) if it["type"] == "score" else None
    return {"ok": top == want, "conf": p[top], "margin": None,
            "brier": sum((a - b) ** 2 for a, b in zip(p, g)),
            "kl": sum(b * math.log(b / max(a, 1e-9)) for a, b in zip(p, g) if b > 0),
            "mae": mae, "type": it["type"], "source": it["source"], "family": it["family"]}
```

## Evaluation

Batches by token budget (long items alone), options in their given order. The decision is the head's for C1 and both averaged for C2; the measures overall and by split, type and family, each readout's measures, and the time per item.

```python
@torch.no_grad()
def evaluate(model, items, max_len, name, pointer):
    model.eval()
    lengths = [len(sequence(it, max_len)[0]) for it in items]
    order = sorted(range(len(items)), key=lengths.__getitem__)
    rows, cut, start, t0 = {}, 0, 0, time.perf_counter()
    while start < len(order):
        batch = [order[start]]
        while (start + len(batch) < len(order)
               and lengths[order[start + len(batch)]] * (len(batch) + 1) <= EVAL_TOKEN_BUDGET):
            batch.append(order[start + len(batch)])
        group = [items[i] for i in batch]
        with amp():
            scores, _, c = forward(model, group, max_len, pointer)
        cut += c
        for readout, probs in readouts(scores).items():
            out = rows.setdefault(readout, [None] * len(items))
            for i, it, p in zip(batch, group, probs.tolist()):
                out[i] = row_of(it, p[:len(it["gold"])])
        start += len(batch)
    ms = (time.perf_counter() - t0) / len(items) * 1000
    main = rows["both" if pointer else "head"]
    result = {"all": measure(main), "readouts": {k: measure(v) for k, v in rows.items()},
              "ms_per_item": round(ms, 1), "states_cut": cut}
    for key in ("source", "type", "family"):
        values = sorted({r[key] for r in main})
        result[key] = {v: measure([r for r in main if r[key] == v]) for v in values}
    log(f"{name}: {result['all']} | {result['ms_per_item']} ms/item | states cut {cut}")
    if pointer:
        log("   readouts: " + " | ".join(f"{k} {v['acc']}" for k, v in result["readouts"].items()))
    for key in ("source", "type"):
        log("   " + " | ".join(f"{k} {v['acc']}" for k, v in result[key].items()))
    return result
```

## Optimizer, schedule and EMA

The adapters at `LR_LORA`, the MLM transform at `LR_ENCODER`, the predictor at `LR_PREDICTOR`, the label head and τ at `LR_HEAD`; warm-up then cosine decay. After every step the target's adapters move towards the encoder's (the frozen ModernBERT is shared by both).

```python
def make_optimizer(model, total_steps):
    groups = [{"params": [p for p in model.encoder.parameters() if p.requires_grad], "lr": LR_LORA},
              {"params": list(model.head.parameters()), "lr": LR_ENCODER},
              {"params": list(model.predictor.parameters()), "lr": LR_PREDICTOR},
              {"params": list(model.label_head.parameters()) + [model.log_tau], "lr": LR_HEAD}]
    opt = torch.optim.AdamW(groups, weight_decay=0.01)
    warm = max(1, int(WARMUP * total_steps))

    def factor(s):
        if s < warm:
            return (s + 1) / warm
        return 0.5 * (1 + math.cos(math.pi * (s - warm) / max(1, total_steps - warm)))

    return opt, torch.optim.lr_scheduler.LambdaLR(opt, factor)


def lora_pairs(model):
    """(target adapter, encoder adapter) parameter pairs."""
    return [(pt, ps) for (name, pt), ps in zip(model.target.named_parameters(),
                                               model.encoder.parameters()) if "lora_" in name]


@torch.no_grad()
def ema_update(model, step, total_steps):
    m = EMA_END - (EMA_END - EMA_START) * (math.cos(math.pi * step / total_steps) + 1) / 2
    for pt, ps in lora_pairs(model):
        pt.mul_(m).add_(ps.detach(), alpha=1 - m)
    return m
```

## Training

A plan is a list of batches (`ARM_EPOCHS` passes, reshuffled each pass); training runs a slice of it, with choice options shuffled per item, and calls `on_step` every `CKPT_EVERY` steps (the checkpoint). Every 50 steps: the decision loss averaged over the last 50 batches, the current batch's losses, the adapters' learning rate, peak GPU memory and host RAM; with the pointer on, cos(ẑ, gold label vector) against the best other option's, how alike a question's label vectors are, τ, ẑ's norm and the EMA momentum.

```python
def make_plan(n, epochs, seed):
    """Batches of item indices: `epochs` passes over n items, reshuffled each pass."""
    rng, plan = random.Random(seed), []
    for _ in range(epochs):
        order = list(range(n))
        rng.shuffle(order)
        plan += [order[b:b + BATCH] for b in range(0, n, BATCH)]
    return plan


def memory():
    gpu = max(torch.cuda.max_memory_allocated(i) for i in range(N_GPU)) / 2 ** 30 if AMP else 0
    ram = f" RAM {psutil.virtual_memory().percent:.0f}%" if psutil else ""
    return f"GPU peak {gpu:.1f} GB{ram}"


@torch.no_grad()
def pointer_stats(model, extra, items):
    cos, valid = extra["cos"], extra["valid"]
    rows = torch.arange(len(items), device=cos.device)
    gold = torch.tensor([max(range(len(it["gold"])), key=it["gold"].__getitem__) for it in items],
                        device=cos.device)
    other = cos.masked_fill(~valid, -2).scatter(1, gold[:, None], -2).max(1).values
    t = F.normalize(extra["targets"], dim=-1)
    k = cos.shape[1]
    pairs = valid[:, :, None] & valid[:, None, :] & ~torch.eye(k, dtype=torch.bool,
                                                               device=cos.device)
    alike = (t @ t.transpose(1, 2))[pairs].mean()
    return (f" | cos gold {cos[rows, gold].mean().item():.3f} best other {other.mean().item():.3f}"
            f" | label vectors alike {alike.item():.3f} | tau {model.tau().item():.3f}"
            f" | z norm {extra['z'].norm(dim=-1).mean().item():.1f}")


def train_stage(model, opt, sched, state, items, name, pointer, plan, on_step=None):
    model.train()
    recent = []
    rng = random.Random(SEED + len(items) + state["step"])
    t0 = time.perf_counter()
    for chunk in plan:
        batch = [shuffled(items[i], rng) for i in chunk]
        with amp():
            scores, extra, _ = forward(model, batch, TRAIN_MAX_LEN, pointer)
        loss, parts = losses(scores, extra, batch)
        scaler = state["scaler"]
        opt.zero_grad(set_to_none=True)
        scaler.scale(loss).backward()
        scaler.unscale_(opt)
        params = [p for g in opt.param_groups for p in g["params"]]
        norm = torch.nn.utils.clip_grad_norm_(params, 1.0)
        scaler.step(opt)
        scaler.update()
        sched.step()
        state["step"] += 1
        m = ema_update(model, state["step"], state["total"]) if pointer else 0.0
        recent = (recent + [parts["dec"]])[-50:]
        if state["step"] % 50 == 0:
            terms = " ".join(f"{k} {v:.4f}" for k, v in parts.items())
            more = pointer_stats(model, extra, batch) + f" | ema {m:.5f}" if pointer else ""
            top_lr = opt.param_groups[0]["lr"]
            log(f"{name} step {state['step']}/{state['total']} | decision loss, last 50 "
                f"{statistics.mean(recent):.4f} | batch: {terms} | grad {norm:.2f} | "
                f"lr {top_lr:.1e} | {memory()}{more}")
        if on_step and state["step"] % CKPT_EVERY == 0:
            on_step()
    log(f"{name}: {len(plan)} steps in {(time.perf_counter() - t0) / 60:.1f} min")
```

## Checkpoints

One file per arm in `CKPT_DIR`: the whole model (including the frozen ModernBERT twice, simplest to restore), the optimizer, the schedule, the loss scaler, the step and the curve so far. It is written to a temporary file first and then renamed, so a crash while saving leaves the previous checkpoint intact. When an arm finishes, its results go to `<arm>_done.json`, its final weights (without the frozen ModernBERT) to `<arm>_final.pt`, and its checkpoint is deleted.

```python
def ckpt_path(name):
    return os.path.join(CKPT_DIR, f"{name}.pt")


def save_ckpt(name, model, opt, sched, state, curve):
    tmp = ckpt_path(name) + ".tmp"
    torch.save({"model": model.state_dict(), "opt": opt.state_dict(), "sched": sched.state_dict(),
                "scaler": state["scaler"].state_dict(), "step": state["step"], "curve": curve}, tmp)
    os.replace(tmp, ckpt_path(name))
    log(f"{name}: checkpoint at step {state['step']}")


def load_ckpt(name, model, opt, sched, state):
    """Restore an arm from its checkpoint -> its curve so far, or None if there is none."""
    if not os.path.exists(ckpt_path(name)):
        return None
    ck = torch.load(ckpt_path(name), map_location="cpu", weights_only=False)
    model.load_state_dict(ck["model"])
    opt.load_state_dict(ck["opt"])
    sched.load_state_dict(ck["sched"])
    state["scaler"].load_state_dict(ck["scaler"])
    state["step"] = ck["step"]
    return ck["curve"]
```

## JevBench's CLI against this kernel

JevBench at the pinned commit, its public items, and a server on `127.0.0.1:PORT` answering `POST /v1/systemone` (the `typesafe` adapter's format) with whichever model and readout `SERVED` holds. `jevbench_cli` checks one item of each type through JevBench's own adapter and scorer, runs the CLI over all 231 public items, summarises them overall and per tier, and estimates Intelligence from the board's published rules (as in `06_jevbench_cli.ipynb`): chance-corrected competence per request type (choice against a random option; noul against 50%, with P(yes) inside (0.2, 0.8) an abstention and so wrong; score against always guessing the middle level), the mean over types per tier clipped to 0–100, tiers weighted easy 0.1, standard 0.2 (the public `original` items), hard 0.4. As answered, and with every noul answer made decisive. Public items only: not a board number.

```python
if not os.path.isdir(JEVBENCH_DIR):
    subprocess.run(["git", "clone", "-q", "https://github.com/fstandhartinger/jevbench.git",
                    JEVBENCH_DIR], check=True)
subprocess.run(["git", "-C", JEVBENCH_DIR, "fetch", "-q", "origin"], check=False)
subprocess.run(["git", "-C", JEVBENCH_DIR, "checkout", "-q", JEVBENCH_COMMIT], check=True)
head = subprocess.run(["git", "-C", JEVBENCH_DIR, "rev-parse", "HEAD"],
                      capture_output=True, text=True, check=True).stdout.strip()
assert head == JEVBENCH_COMMIT, head
TASK_FILES = {tier: f"{JEVBENCH_DIR}/datasets/public/{tier}.jsonl" for tier in TIERS}
ALL_TASKS = ",".join(TASK_FILES.values())
sys.path.insert(0, JEVBENCH_DIR)
from jevbench.adapters.typesafe import TypeSafeAdapter
from jevbench.scoring import score_task
from jevbench.tasks import load_jsonl

jb_tasks = {tier: load_jsonl(path) for tier, path in TASK_FILES.items()}
tier_of = {t.id: tier for tier, ts in jb_tasks.items() for t in ts}
by_id = {t.id: t for ts in jb_tasks.values() for t in ts}
SERVED = {"model": None, "readout": "both", "name": "", "pointer": True}
print(f"JevBench at {head[:7]} | {len(by_id)} public items")


@torch.no_grad()
def decide(state, q):
    names, texts = option_texts(q)
    if len(names) > MAX_OPTIONS:
        raise ValueError(f"{len(names)} options, at most {MAX_OPTIONS}")
    item = {"type": q["type"], "question": q.get("instructions") or "", "state": state_text(state),
            "names": names, "texts": texts, "gold": [1 / len(names)] * len(names)}
    with amp():
        scores, _, _ = forward(SERVED["model"], [item], EVAL_MAX_LEN, SERVED["pointer"])
    p = readouts(scores)[SERVED["readout"]][0, :len(names)].double()
    return names, (p / p.sum()).tolist(), len(sequence(item, EVAL_MAX_LEN)[0])


def answer(body):
    q = body["questions"]["decision"]
    names, p, n_tokens = decide(body["state"], q)
    if q["type"] == "noul":
        out = {"type": "noul", "noul": p[0]}
    elif q["type"] == "choice":
        out = {"type": "choice", "choice": names[max(range(len(p)), key=p.__getitem__)],
               "probabilities": dict(zip(names, p))}
    else:
        out = {"type": "score", "score": sum(i * x for i, x in enumerate(p)),
               "probabilities": {str(i): x for i, x in enumerate(p)}}
    return {"model": SERVED["name"], "answers": {"decision": out},
            "usage": {"input_tokens": n_tokens, "output_tokens": 0}}


class Handler(BaseHTTPRequestHandler):
    def do_POST(self):
        if self.path != "/v1/systemone":
            return self.reply(404, {"error": "not found"})
        try:
            body = json.loads(self.rfile.read(int(self.headers["Content-Length"])))
            self.reply(200, answer(body))
        except Exception as e:
            self.reply(400, {"error": f"{type(e).__name__}: {e}"})

    def reply(self, status, obj):
        data = json.dumps(obj).encode("utf-8")
        self.send_response(status)
        self.send_header("Content-Type", "application/json")
        self.send_header("Content-Length", str(len(data)))
        self.end_headers()
        self.wfile.write(data)

    def log_message(self, *args):
        pass


old_server = globals().get("server")
if old_server is not None:                         # the cell run again: stop the old one first
    old_server.shutdown()
    old_server.server_close()
server = HTTPServer(("127.0.0.1", PORT), Handler)
threading.Thread(target=server.serve_forever, daemon=True).start()
log(f"server at http://127.0.0.1:{PORT}/v1/systemone")


def cli_summary(task_files, results):
    out = subprocess.run([sys.executable, "-m", "jevbench.cli", "summarize", "--tasks", task_files,
                          "--results", results],
                         cwd=JEVBENCH_DIR, capture_output=True, text=True, check=True)
    s = json.loads(out.stdout)
    return {"n": s["n_scorable"], "acc": s["accuracy"], "brier": s["brier_mean"],
            "ece": s["ece"]["ece"], "ordinal_mae": s["ordinal_mae"], "valid": s["n_valid"],
            "p50_ms": s["latency"]["p50_s"] * 1000, "p95_ms": s["latency"]["p95_s"] * 1000}


def competence(rs, kind, decisive_all):
    ts = [by_id[r["task_id"]] for r in rs]
    if kind == "choice":
        acc = statistics.mean(bool(r["correct"]) for r in rs)
        base = statistics.mean(1 / len(t.labels) for t in ts)
        return 100 * (acc - base) / (1 - base), None
    if kind == "noul":
        hits, decisive = [], []
        for r in rs:
            p = (r.get("probs") or {}).get("yes")
            sure = p is not None and (decisive_all or p >= NOUL_DECISIVE or p <= 1 - NOUL_DECISIVE)
            decisive.append(sure)
            hits.append(sure and bool(r["correct"]))
        return 100 * (statistics.mean(hits) - 0.5) / 0.5, statistics.mean(decisive)
    errs, base = [], []
    for r, t in zip(rs, ts):
        mid = (len(t.labels) - 1) / 2
        ev = r.get("ordinal_ev")
        errs.append(abs(ev - t.expected) if ev is not None else abs(len(t.labels) - 1))
        base.append(abs(mid - t.expected))
    if statistics.mean(base) == 0:
        base = [statistics.mean(abs(i - t.expected) for i in range(len(t.labels))) for t in ts]
    return 100 * (statistics.mean(base) - statistics.mean(errs)) / statistics.mean(base), None


def intelligence(records, decisive_all):
    tiers, total = {}, 0.0
    for tier, weight in TIERS.items():
        per = {}
        for kind in TYPES:
            rs = [r for r in records if tier_of[r["task_id"]] == tier
                  and by_id[r["task_id"]].question["type"] == kind]
            if rs:
                per[kind] = competence(rs, kind, decisive_all)
        o1s = min(100.0, max(0.0, statistics.mean(c for c, _ in per.values())))
        tiers[tier] = {"o1s": o1s, "types": per}
        total += weight * o1s
    return {"intelligence": total / sum(TIERS.values()), "tiers": tiers}


def jevbench_cli(model, name, readout, pointer=True):
    """Serve `model` with `readout` and score it with JevBench's CLI on the public items."""
    model.eval()
    SERVED.update(model=model, readout=readout, name=f"{name}-{readout}", pointer=pointer)
    adapter = TypeSafeAdapter(endpoint=f"http://127.0.0.1:{PORT}", model=SERVED["name"], key_env="")
    for kind in TYPES:                                # one item of each type, the CLI's own way
        t = next(t for t in by_id.values() if t.question["type"] == kind)
        r = adapter.run(t)
        assert r.ok and score_task(r.probs, t)["valid"], (kind, r.error)
    run_dir = f"{RUNS_DIR}/{SERVED['name']}-{time.strftime('%Y%m%d-%H%M%S')}"
    os.makedirs(run_dir)
    cmd = [sys.executable, "-m", "jevbench.cli", "run", "--tasks", ALL_TASKS, "--adapter", "typesafe",
           "--endpoint", f"http://127.0.0.1:{PORT}", "--model", SERVED["name"], "--key-env", "",
           "--results", f"{run_dir}/results.jsonl", "--ledger", f"{run_dir}/ledger.jsonl",
           "--raw-dir", f"{run_dir}/raw", "--reserve-usd", "0", "--price-in-per-m", "0",
           "--price-out-per-m", "0", "--cost-basis", "local_gpu_no_billable_account",
           "--run-label", SERVED["name"], "--manifest", f"{run_dir}/manifest.json"]
    done = subprocess.run(cmd, cwd=JEVBENCH_DIR, capture_output=True, text=True)
    assert done.returncode == 0, done.stdout[-1500:] + done.stderr[-1500:]
    lines = open(f"{run_dir}/results.jsonl").read().splitlines()
    out = {"run_dir": run_dir, "cli": {"all": cli_summary(ALL_TASKS, f"{run_dir}/results.jsonl")}}
    for tier, path in TASK_FILES.items():
        wanted = {json.loads(line)["id"] for line in open(path)}
        part = f"{run_dir}/results-{tier}.jsonl"           # the summariser wants one file's items
        with open(part, "w") as fh:
            fh.writelines(line + "\n" for line in lines if json.loads(line)["task_id"] in wanted)
        out["cli"][tier] = cli_summary(path, part)
    records = [json.loads(line) for line in lines]
    out["as answered"] = intelligence(records, False)
    out["noul decisive"] = intelligence(records, True)
    a = out["cli"]["all"]
    log(f"{SERVED['name']} JevBench CLI: acc {a['acc']:.3f} brier {a['brier']:.3f} ece {a['ece']:.3f}"
        f" | p50 {a['p50_ms']:.0f} ms | Intelligence (est.) {out['as answered']['intelligence']:.1f}"
        f" as answered, {out['noul decisive']['intelligence']:.1f} noul decisive")
    return out
```

## Checks before training

For each predictor in the arms, on a small batch from every question type: ModernBERT's own weights are frozen and only adapters train in the encoder; the target gets no gradients and stays in eval mode after `train()`; its outputs are deterministic; every loss is finite; the answer slot holds only `[MASK]`; each label position holds that label's token; the EMA step moves the target's adapters towards the encoder's; a checkpoint saved and loaded restores the weights. Then the parameter counts.

```python
def checks(predictor):
    model = MaskedDecider(predictor).to(DEVICE)
    sample = [next(it for it in train_items if it["type"] == k) for k in TYPES]
    sample += [next(it for it in ts_items if it["type"] == k) for k in TYPES]
    model.train()
    assert not any(p.requires_grad for n, p in model.encoder.named_parameters() if "lora_" not in n)
    assert not model.target.training, "target encoder left in training mode"
    assert not any(p.requires_grad for p in model.target.parameters())
    with amp():
        scores, extra, _ = forward(model, sample, TRAIN_MAX_LEN, pointer=True)
        again = forward(model, sample, TRAIN_MAX_LEN, pointer=True)[1]["targets"]
    assert torch.allclose(extra["targets"], again), "target outputs are not deterministic"
    loss, parts = losses(scores, extra, sample)
    assert all(math.isfinite(v) for v in parts.values()), parts
    loss.backward()
    assert all(p.grad is None for p in model.target.parameters()), "target received gradients"
    for it in sample:
        seq, pos, at, _ = sequence(it, TRAIN_MAX_LEN)
        assert seq[pos] == MASK and seq[pos + 1:] == [SEP], "the answer slot is not empty"
        for label, a in zip(LABELS[it["type"]], at):
            assert seq[a] == ids(label)[-1], f"wrong position for label {label!r}"
    p_t, p_o = lora_pairs(model)[0]
    with torch.no_grad():
        p_o.add_(0.01)
    before = (p_t - p_o).abs().sum().item()
    ema_update(model, 1, 10)
    after = (p_t - p_o).abs().sum().item()
    assert after < before, "EMA did not move the target's adapters towards the encoder's"
    opt, sched = make_optimizer(model, 10)
    state = {"step": 3, "total": 10, "scaler": new_scaler()}
    save_ckpt("check", model, opt, sched, state, ["curve"])
    kept = p_o.detach().clone()
    with torch.no_grad():
        p_o.add_(1.0)
    state["step"] = 0
    curve = load_ckpt("check", model, opt, sched, state)
    assert torch.equal(p_o, kept) and state["step"] == 3 and curve == ["curve"], "checkpoint"
    os.remove(ckpt_path("check"))
    count = lambda ps: sum(p.numel() for p in ps) / 1e6
    adapters = count(p for p in model.encoder.parameters() if p.requires_grad)
    log(f"checks passed ({predictor}) | parts {parts} | ModernBERT "
        f"{count(model.encoder.parameters()) - adapters:.1f} M frozen | adapters {adapters:.1f} M | "
        f"MLM transform + label head {count(model.head.parameters()) + count(model.label_head.parameters()):.1f} M"
        f" | predictor {count(model.predictor.parameters()):.1f} M | final weights "
        f"{sum(v.numel() * v.element_size() for v in model.final_state().values()) / 2 ** 20:.0f} MB")
    del model, opt
    gc.collect()
    torch.cuda.empty_cache()


for predictor in sorted({PREDICTORS[a] or "mlp" for a in ARMS}):
    checks(predictor)
```

## One arm, with a learning curve and checkpoints

From the pretrained weights, or from the arm's checkpoint: one learning-rate schedule over `ARM_EPOCHS` passes; at each quarter, 1,000 of its own training questions, our dev and JevBench. A quarter already scored before a crash is not trained or scored again.

```python
def point(model, name, label, seen, pointer):
    """(label, training questions, dev, JevBench) at one point of the curve."""
    return (label,
            evaluate(model, seen, TRAIN_MAX_LEN, f"{name} train, {label}", pointer),
            evaluate(model, dev_items, TRAIN_MAX_LEN, f"{name} dev, {label}", pointer),
            evaluate(model, jev_items, EVAL_MAX_LEN, f"{name} JevBench, {label}", pointer))


def run(name, items, pointer, predictor):
    random.seed(SEED)
    torch.manual_seed(SEED)
    model = MaskedDecider(predictor).to(DEVICE)
    seen = random.Random(SEED).sample(items, 1000)
    plan = make_plan(len(items), ARM_EPOCHS, SEED + len(items))
    opt, sched = make_optimizer(model, len(plan))
    state = {"step": 0, "total": len(plan), "scaler": new_scaler()}
    curve = load_ckpt(name, model, opt, sched, state)
    resumed = curve is not None
    if resumed:
        log(f"{name}: resumed at step {state['step']}/{len(plan)}, "
            f"{len(curve) - 1} of {CURVE_POINTS} quarters scored")
    else:
        curve = [point(model, name, "untrained", seen, pointer)]

    def save():
        save_ckpt(name, model, opt, sched, state, curve)

    if not resumed:
        save()
    cuts = [round(len(plan) * k / CURVE_POINTS) for k in range(CURVE_POINTS + 1)]
    for k in range(CURVE_POINTS):
        if len(curve) > k + 1:
            continue
        train_stage(model, opt, sched, state, items, name, pointer,
                    plan[max(cuts[k], state["step"]):cuts[k + 1]], on_step=save)
        curve.append(point(model, name, f"{(k + 1) * ARM_EPOCHS / CURVE_POINTS:g} passes", seen,
                           pointer))
        save()
    return model, curve
```

## Typed Decisions' train split (never trained on)

Each arm is scored on it at the end. For reference: our 4B decoder 0.720, plain ModernBERT-base (run 8 B) 0.350, run 11's C2 0.332 (pointer alone 0.383), chance 0.318.

```python
td_path = hf_hub_download("LocalLLaMA/typed-decisions", "all/train-00000-of-00001.parquet",
                          repo_type="dataset")
td_items = []
for row in pd.read_parquet(td_path).to_dict("records"):
    state, qs, gold = (json.loads(row[k]) for k in ("state", "questions", "gold"))
    for qid, q in qs.items():
        names, _ = option_texts(q)
        probs = gold[qid]["probabilities"]
        if len(names) <= MAX_OPTIONS and all(n in probs for n in names):
            td_items.append(make_item(q["type"], q.get("instructions"), state, q,
                                      [float(probs[n]) for n in names], row["workflow"],
                                      f"{row['workflow']}/{qid}"))
print(f"Typed Decisions train: {len(td_items)} questions")
```

## The arms

L3 (transformer predictor), then L2 (MLP predictor), both with the LoRA encoder and the pointer. After training: Typed Decisions' train split, then JevBench's CLI with the averaged decision and with the head alone. A finished arm's results are read from `CKPT_DIR`; an unfinished arm resumes from its checkpoint.

```python
mixed = ts_items + train_items
curves, td_scores, cli_scores = {}, {}, {}
for name in ARMS:
    done = os.path.join(CKPT_DIR, f"{name}_done.json")
    if os.path.exists(done):
        saved = json.load(open(done))
        curves[name], td_scores[name], cli_scores[name] = saved["curve"], saved["td"], saved["cli"]
        log(f"{name}: finished earlier, results read from {done}")
        continue
    pointer = PREDICTORS[name] is not None
    model, curves[name] = run(name, mixed, pointer, PREDICTORS[name] or "mlp")
    td_scores[name] = evaluate(model, td_items, TRAIN_MAX_LEN, f"{name} on Typed Decisions train",
                               pointer)
    torch.save(model.final_state(), os.path.join(CKPT_DIR, f"{name}_final.pt"))
    cli_scores[name] = {readout: jevbench_cli(model, name, readout, pointer)
                        for readout in (("both", "head") if pointer else ("head",))}
    json.dump({"curve": curves[name], "td": td_scores[name], "cli": cli_scores[name]}, open(done, "w"))
    os.remove(ckpt_path(name))
    SERVED["model"] = None
    del model
    gc.collect()
    torch.cuda.empty_cache()
```

## The comparison

Each arm's learning curve with each readout, the final scores next to run 8 B and run 11's C2 (both full fine-tuning of ModernBERT-base with the MLP predictor), and JevBench's CLI with our Intelligence estimate.

```python
def chance(items):
    return round(statistics.mean(1 / len(it["names"]) for it in items), 3)


def accs(result):
    return " ".join(f"{k} {v['acc']:.3f}" for k, v in result["readouts"].items())


for name, curve in curves.items():
    print(f"{name} ({PREDICTORS[name] or 'no JEPA'}){'point':>11s} {'train':>7s} {'dev':>7s} {'JevBench':>9s} "
          f"{'original':>9s} {'easy':>7s} {'hard':>7s}")
    for where, tr, dv, jv in curve:
        splits = " ".join(f"{jv['source'][s]['acc']:>7.3f}" for s in ("original", "easy", "hard"))
        print(f"{'':16s}{where:>11s} {tr['all']['acc']:>7.3f} {dv['all']['acc']:>7.3f} "
              f"{jv['all']['acc']:>9.3f}  {splits}")
        print(f"{'':28s}dev: {accs(dv)} | JevBench: {accs(jv)}")
print(f"\n{'final':16s} {'JevBench':>9s} {'easy':>6s} {'dev':>6s} {'TD train':>9s}")
print(f"{'run 8 B':16s} {0.394:>9.3f} {0.562:>6.3f} {0.457:>6.3f} {0.350:>9.3f}   (head only, 2 passes)")
print(f"{'run 11 C2':16s} {0.489:>9.3f} {0.854:>6.3f} {0.491:>6.3f} {'-':>9s}   (1 pass)")
print(f"{'run 11 C2':16s} {0.493:>9.3f} {0.833:>6.3f} {0.471:>6.3f} {0.332:>9.3f}   (2 passes)")
for name, curve in curves.items():
    jv, dv, td = curve[-1][3], curve[-1][2], td_scores[name]
    print(f"{name + ' ' + (PREDICTORS[name] or 'no JEPA'):16s} {jv['all']['acc']:>9.3f} "
          f"{jv['source']['easy']['acc']:>6.3f} {dv['all']['acc']:>6.3f} {td['all']['acc']:>9.3f}")
    print(f"{'':16s} JevBench: {accs(jv)} | dev: {accs(dv)} | TD: {accs(td)}")
print(f"chance: JevBench {chance(jev_items)} | dev {chance(dev_items)} | TD {chance(td_items)}")

print(f"\nJevBench CLI, public items{'':6s}{'acc':>6s} {'brier':>6s} {'ece':>6s} {'p50 ms':>7s} "
      f"{'Intel. as answered':>19s} {'noul decisive':>14s}")
for name, by_readout in cli_scores.items():
    for readout, out in by_readout.items():
        a = out["cli"]["all"]
        print(f"{name + ' ' + readout:32s}{a['acc']:>6.3f} {a['brier']:>6.3f} {a['ece']:>6.3f} "
              f"{a['p50_ms']:>7.0f} {out['as answered']['intelligence']:>19.1f} "
              f"{out['noul decisive']['intelligence']:>14.1f}")
        for tier in TIERS:
            t = out["as answered"]["tiers"][tier]
            parts = " | ".join(f"{k} {c:.1f}" + (f" (decisive {d:.0%})" if d is not None else "")
                               for k, (c, d) in t["types"].items())
            print(f"{'':6s}{tier:9s} O1S {t['o1s']:5.1f} | {parts}")
```
